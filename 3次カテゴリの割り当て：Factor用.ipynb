```python
# -*- coding: utf-8 -*-
# Google Colabで実行することを想定

# 1. 必要なライブラリをインストール
# gspread: Google Sheets APIを操作
# pandas: データ操作
# google-generativeai: Gemini APIを利用
# Google認証関連ライブラリ
#!pip install --upgrade google-auth google-auth-oauthlib google-auth-httplib2 google-api-python-client gspread pandas google-generativeai -q
#
#print("ライブラリのインストールが完了しました。")

# 2. Google Driveのマウントと認証 (スプレッドシートアクセスに必要)
from google.colab import auth
try:
    auth.authenticate_user()
    print("Googleアカウント認証が成功しました。")
except Exception as e:
    print(f"Googleアカウント認証に失敗しました: {e}")
    print("ノートブックの設定でサードパーティのCookieが許可されているか確認してください。")
    raise # エラーで停止

import gspread
from google.auth import default
import re
import pandas as pd
import google.generativeai as genai # Gemini APIライブラリ
import time # API呼び出し間の遅延のため
# === ▼▼▼ コンテキストキャッシュのために追加 ▼▼▼ ===
import google.generativeai.caching as caching
import datetime
# === ▲▲▲ コンテキストキャッシュのために追加 ▲▲▲ ===
# === ▼▼▼ リトライ処理のために追加 ▼▼▼ ===
from google.api_core.exceptions import RetryError, Aborted, Cancelled, DeadlineExceeded, ServiceUnavailable, Unknown # より広範なリトライ対象エラー
import requests # requests.exceptions.ConnectionError のため
# === ▲▲▲ リトライ処理のために追加 ▲▲▲ ===


# Google APIクライアントの認証情報を取得 (gspread用)
try:
    creds, _ = default()
    gc = gspread.authorize(creds)
    print("Google Sheets APIへのアクセス準備ができました。")
except Exception as e:
    print(f"Google Sheets APIの認証に失敗しました: {e}")
    raise # エラーで停止

# 3. スプレッドシート情報、三次カテゴリリスト、Gemini APIキー設定
# --------------------------------------------------
# ▼▼▼ ユーザー設定箇所 ▼▼▼

# 使用するGeminiモデル (プルダウンから選択)
# 注意: コンテキストキャッシュは 'gemini-1.5-pro-latest' や 'gemini-1.5-flash-latest' などの安定版モデルで動作確認されています。
# 'gemini-2.5' などのプレビュー版モデルでは、キャッシュ作成時にエラーが発生する場合があります。
GEMINI_MODEL_NAME = "gemini-2.5-pro" #@param ["gemini-2.5-pro", "gemini-2.5-flash"]
print(f"使用するGeminiモデル: {GEMINI_MODEL_NAME}")

# 対象のスプレッドシートURL (必ず編集してください)
SPREADSHEET_URL = 'https://docs.google.com/spreadsheets/d/1bFlIif-Uu7Df2jdSFqY8uqLbLDE36zvz1jFOvmD-6QI/edit?gid=1515956011#gid=1515956011' #@param {type:"string"}
print(f"指定されたスプレッドシートURL: {SPREADSHEET_URL}")

# 対象のシート名 (Colabのフォームで指定)
SHEET_NAME = "シート1" #@param {type:"string"}
print(f"処理対象のシート名: {SHEET_NAME}")

# 割り当てる三次カテゴリの最大数 (Colabのフォームで指定)
MAX_TERTIARY_CATEGORIES_TO_ASSIGN = 5 #@param {type:"integer"}
print(f"割り当てる三次カテゴリの最大数: {MAX_TERTIARY_CATEGORIES_TO_ASSIGN}")

# キーワードが記載されている列 (例: 'A')
KEYWORD_COLUMN = "A" #@param {type:"string"}
print(f"キーワード列: {KEYWORD_COLUMN}")

# 割り当てられた三次カテゴリを書き込む列 (例: 'B')
ASSIGNED_TERTIARY_CATEGORY_COLUMN = "B" #@param {type:"string"}
print(f"割り当て三次カテゴリ書き込み列: {ASSIGNED_TERTIARY_CATEGORY_COLUMN}")

# 提案/エラー情報を書き込む列 (例: 'C')
SUGGESTION_COLUMN = "C" #@param {type:"string"}
print(f"提案/エラー情報書き込み列: {SUGGESTION_COLUMN}")

# === ▼▼▼ 書き込みバッチサイズ設定 ▼▼▼ ===
# Google Sheets APIの書き込みクォータ制限を回避するため、一度に書き込む行数を設定します。
BATCH_SIZE = 10 #@param {type:"integer"}
print(f"スプレッドシートへの書き込みバッチサイズ: {BATCH_SIZE} 行")
# === ▲▲▲ 書き込みバッチサイズ設定 ▲▲▲ ===


# === ▼▼▼ リトライ設定 ▼▼▼ ===
MAX_API_RETRIES = 3  # API呼び出しの最大リトライ回数
API_RETRY_DELAY_SECONDS = 5  # リトライ間の待機時間（秒）
# === ▲▲▲ リトライ設定 ▲▲▲ ===

# 関連性を判定したい三次カテゴリのリスト (自由に編集・追加してください)
# `categor_tertiary_original` は colab 画面上では1つ前のセルに記載されていることを想定

# === ▼▼▼ 除外したい三次カテゴリをここに記述 ▼▼▼ ===
categories_to_exclude = [
    'Electrification'
]
# === ▲▲▲ 除外したい三次カテゴリをここに記述 ▲▲▲ ===

# 元のリストから除外リストに含まれる要素を除去
tertiary_categories = [sc for sc in categor_tertiary_original if sc not in categories_to_exclude]
print(f"元の三次カテゴリ数: {len(categor_tertiary_original)}")
print(f"除外後の三次カテゴリ数: {len(tertiary_categories)}")
if len(categories_to_exclude) > 0:
    print(f"除外されたカテゴリ: {', '.join(categories_to_exclude)}")

# ▼▼▼ Gemini APIキー設定 (ユーザーが必ず設定してください) ▼▼▼
from google.colab import userdata
print("Colabシークレットの準備を待っています...")
time.sleep(1)
print("シークレットの待機完了。APIキーの読み込みを試みます。")
try:
    GEMINI_API_KEY = userdata.get('GEMINI_API_KEY')
    genai.configure(api_key=GEMINI_API_KEY)
    print("Gemini APIキーがColabシークレットから正常に読み込まれました。")
except userdata.SecretNotFoundError:
    print("エラー: Colabのシークレットに 'GEMINI_API_KEY' が見つかりません。")
    raise ValueError("Gemini APIキーが設定されていません。処理を続行できません。")
except Exception as e:
    print(f"Gemini APIキーの設定中にエラーが発生しました: {e}")
    raise

# APIキー設定と指定されたモデルの存在確認
try:
    print(f"指定されたモデル '{GEMINI_MODEL_NAME}' を使用してGenerativeModelを初期化します。")
    test_model_init = genai.GenerativeModel(GEMINI_MODEL_NAME)
    print(f"モデル '{GEMINI_MODEL_NAME}' (実名: {test_model_init.model_name}) は利用可能です。")
except Exception as e:
    print(f"警告: 指定されたモデル '{GEMINI_MODEL_NAME}' の初期化中にエラーが発生しました: {e}")

# ▲▲▲ ユーザー設定箇所 ▲▲▲
# --------------------------------------------------

# --- 処理開始 ---

# スプレッドシートIDをURLから抽出
spreadsheet_id = None
if SPREADSHEET_URL:
    match = re.search(r'/d/([a-zA-Z0-9-_]+)', SPREADSHEET_URL)
    if match:
        spreadsheet_id = match.group(1)
        print(f"スプレッドシートID: {spreadsheet_id} を抽出しました。")
    else:
        raise ValueError("エラー: 有効なGoogle スプレッドシートのURL形式ではありません。")
else:
    raise ValueError("エラー: スプレッドシートのURLが指定されていません。")

# 列名を数値インデックスに変換するヘルパー関数
def column_to_index(col_name):
    index = 0
    for char in col_name.upper():
        index = index * 26 + (ord(char) - ord('A') + 1)
    return index

# 6. スプレッドシートを開き、データを読み込む
try:
    print(f"スプレッドシート '{spreadsheet_id}' を開いています...")
    spreadsheet = gc.open_by_key(spreadsheet_id)
    worksheet = spreadsheet.worksheet(SHEET_NAME)
    print(f"シート '{SHEET_NAME}' へのアクセスに成功しました。")

    print("シートの全データを読み込んでいます...")
    all_sheet_data = worksheet.get_all_values()
    if not all_sheet_data:
        print("シートにデータがありません。")
    else:
        print(f"{len(all_sheet_data)} 行のデータを読み込みました。")

        # === ▼▼▼ ステップ1: キャッシュするシステムプロンプトを定義 ▼▼▼ ===
        # `examples_in_markdown` は colab 画面上では1つ前のセルに記載されていることを想定
        additional_examples = "\n具体例のテーブルはここまで。\n"
        examples_in_markdown += additional_examples

        tertiary_categories_str_for_prompt = ", ".join(f"「{sc}」" for sc in tertiary_categories)

        system_prompt = f"""あなたは不動産ESGを専門とするテキスト分類アシスタントです。
提示された「キーワード」に対して、以下の「三次カテゴリリスト」の中から、キーワードの文言から**直接的かつ明確に**示されている内容と合致する、最も中心的な三次カテゴリを最大{MAX_TERTIARY_CATEGORIES_TO_ASSIGN}つまで選択してください。
キーワードが示唆する可能性のある関連施策や、結果として達成されうる具体的な認証などを、キーワード中に明示的な言及がない限り、関連性が高いとは判断しないでください。
キーワードの中心的な意味内容に合致する三次カテゴリを選んでください。
関連性が低い場合や、直接的に合致する三次カテゴリがない場合は無理に選択せず、何も返さないでください。
出力形式は、選択した三次カテゴリ名をカンマ区切りにした文字列のみとし、他の挨拶や説明文は一切含めないでください。

以下は出力の具体例と補足事項です。
{examples_in_markdown}

三次カテゴリリスト:
{tertiary_categories_str_for_prompt}

キーワードを受け取ったら、分類結果のみを出力してください。"""
        # === ▲▲▲ ステップ1 ▲▲▲ ===

        # === ▼▼▼ ステップ2 & 3: キャッシュの作成とモデルの準備（フォールバック付き）▼▼▼ ===
        gemini_model_instance = None
        is_cache_enabled = False
        cached_content_token_count = 0
        successful_api_calls = 0

        try:
            print("\nコンテキストキャッシュの作成を試みています...")
            # キャッシュの有効期限をTTLで設定 (例: 1時間)
            cached_content = caching.CachedContent.create(
                model=f'models/{GEMINI_MODEL_NAME}',
                system_instruction=system_prompt,
                ttl=datetime.timedelta(hours=1),
            )
            gemini_model_instance = genai.GenerativeModel.from_cached_content(cached_content)
            is_cache_enabled = True
            # 節約トークン計算のため、キャッシュしたコンテンツのトークン数を数える
            cached_content_token_count = genai.GenerativeModel(GEMINI_MODEL_NAME).count_tokens(system_prompt).total_tokens
            print(f"コンテキストキャッシュの作成に成功しました。キャッシュ名: {cached_content.name}")
            print(f"キャッシュされたコンテンツのトークン数: {cached_content_token_count}")

        except Exception as e:
            print(f"コンテキストキャッシュの作成に失敗しました: {e}")
            print("原因として、選択されたモデルがコンテキストキャッシュ機能に未対応の可能性があります。")
            print("通常のモデルを初期化して処理を続行します。")
            gemini_model_instance = genai.GenerativeModel(GEMINI_MODEL_NAME)
            is_cache_enabled = False
        # === ▲▲▲ ステップ2 & 3 ▲▲▲ ===

        print(f"\nGemini APIモデル ({GEMINI_MODEL_NAME}) の準備ができました。キーワードの分類を開始します...")

        keyword_col_idx = column_to_index(KEYWORD_COLUMN) - 1
        assigned_col_idx = column_to_index(ASSIGNED_TERTIARY_CATEGORY_COLUMN) - 1
        suggestion_col_idx = column_to_index(SUGGESTION_COLUMN) - 1

        total_input_tokens = 0
        total_output_tokens = 0

        cells_to_update_batch = []

        for i, row_data in enumerate(all_sheet_data):
            current_row_in_sheet = i + 1
            if i == 0: continue
            keyword = row_data[keyword_col_idx] if len(row_data) > keyword_col_idx else ""
            if not keyword or not keyword.strip(): continue

            existing_result = row_data[assigned_col_idx] if len(row_data) > assigned_col_idx else ""
            if existing_result and existing_result.strip():
                print(f"行 {current_row_in_sheet}: 処理済みのためスキップします。")
                continue

            current_keyword_for_log = keyword[:100] + '...' if len(keyword) > 100 else keyword
            print(f"\n処理中 (行 {current_row_in_sheet}): \"{current_keyword_for_log}\"")

            assigned_tertiary_categories_str = ""
            suggestion_or_error_str = ""

            content_to_send = None
            if is_cache_enabled:
                content_to_send = keyword
            else:
                content_to_send = f"{system_prompt}\n\nキーワード:\n{keyword}\n\n出力:"

            response = None
            for attempt in range(MAX_API_RETRIES):
                try:
                    response = gemini_model_instance.generate_content(content_to_send)
                    successful_api_calls += 1
                    break
                except (requests.exceptions.ConnectionError, RetryError, Aborted, Cancelled, DeadlineExceeded, ServiceUnavailable, Unknown) as e:
                    print(f"  API呼び出し試行 {attempt + 1}/{MAX_API_RETRIES} でエラー: {type(e).__name__}")
                    if attempt < MAX_API_RETRIES - 1:
                        time.sleep(API_RETRY_DELAY_SECONDS)
                    else:
                        assigned_tertiary_categories_str = "API接続エラー(リトライ上限)"
                        suggestion_or_error_str = f"API接続エラー ({type(e).__name__})"
                except Exception as e:
                    print(f"  キーワード '{current_keyword_for_log}' の処理中に予期せぬGemini APIエラーが発生しました: {type(e).__name__}")
                    assigned_tertiary_categories_str = "API例外エラー"
                    suggestion_or_error_str = f"APIエラー ({type(e).__name__})"
                    break

            if assigned_tertiary_categories_str:
                pass
            elif response and response.candidates and response.candidates[0].content and response.candidates[0].content.parts:
                try:
                    input_tokens = response.usage_metadata.prompt_token_count
                    output_tokens = response.usage_metadata.candidates_token_count
                    total_input_tokens += input_tokens
                    total_output_tokens += output_tokens
                    print(f"  -> 消費トークン: 入力 {input_tokens}, 出力 {output_tokens}")
                except Exception:
                    print("  -> 消費トークン数の取得に失敗しました。")

                gemini_output_raw = response.text.strip()
                print(f"  Geminiからの生出力: \"{gemini_output_raw}\"")
                if not gemini_output_raw:
                    assigned_tertiary_categories_str = "(関連カテゴリなし)"
                    suggestion_or_error_str = "Geminiにより関連カテゴリなしと判断されました。"
                else:
                    returned_tertiary_categories = [s.strip().replace("「", "").replace("」", "") for s in gemini_output_raw.split(',') if s.strip()]
                    valid_assigned_tertiary_categories = []
                    invalid_returned_categories = []
                    for r_sc in returned_tertiary_categories:
                        if r_sc in tertiary_categories:
                            if len(valid_assigned_tertiary_categories) < MAX_TERTIARY_CATEGORIES_TO_ASSIGN:
                                valid_assigned_tertiary_categories.append(r_sc)
                        else:
                            invalid_returned_categories.append(r_sc)
                    if valid_assigned_tertiary_categories:
                        assigned_tertiary_categories_str = ", ".join(valid_assigned_tertiary_categories)
                    else:
                        assigned_tertiary_categories_str = "(有効なカテゴリなし)"
                    if invalid_returned_categories:
                        suggestion_or_error_str = f"Geminiがリスト外のカテゴリを返しました: {', '.join(invalid_returned_categories)}"
                    elif not valid_assigned_tertiary_categories and not gemini_output_raw:
                         assigned_tertiary_categories_str = "(関連カテゴリなし)"
                         suggestion_or_error_str = "Geminiが関連カテゴリを返しませんでした。"
            elif response:
                error_message = "Geminiから有効な応答構造が得られませんでした。"
                assigned_tertiary_categories_str = "API応答構造エラー"
                suggestion_or_error_str = error_message

            cells_to_update_batch.append(gspread.Cell(current_row_in_sheet, assigned_col_idx + 1, assigned_tertiary_categories_str))
            cells_to_update_batch.append(gspread.Cell(current_row_in_sheet, suggestion_col_idx + 1, suggestion_or_error_str))
            print(f"  -> 結果をバッチに追加: '{assigned_tertiary_categories_str}'")

            # === ▼▼▼ 変更点: バッチ書き込み条件を修正 ▼▼▼ ===
            # バッチサイズに達した場合のみ書き込む
            if len(cells_to_update_batch) >= BATCH_SIZE * 2:
                try:
                    print(f"\n--- バッチ書き込み中 ({len(cells_to_update_batch) // 2} 件) ---")
                    worksheet.update_cells(cells_to_update_batch, value_input_option='USER_ENTERED')
                    print("--- バッチ書き込み完了 ---")
                    cells_to_update_batch.clear() # バッチをクリア
                except Exception as e:
                    print(f"  バッチの書き込み中にエラーが発生しました: {e}")
                    print("30秒待機して処理を続行します...")
                    time.sleep(30)
            # === ▲▲▲ 変更点 ▲▲▲ ===

            time.sleep(1)

        # === ▼▼▼ 変更点: ループ終了後、残りのバッチを書き込む処理を追加 ▼▼▼ ===
        if cells_to_update_batch:
            try:
                print(f"\n--- 最終バッチ書き込み中 ({len(cells_to_update_batch) // 2} 件) ---")
                worksheet.update_cells(cells_to_update_batch, value_input_option='USER_ENTERED')
                print("--- 最終バッチ書き込み完了 ---")
                cells_to_update_batch.clear()
            except Exception as e:
                print(f"  最終バッチの書き込み中にエラーが発生しました: {e}")
        # === ▲▲▲ 変更点 ▲▲▲ ===


except gspread.exceptions.SpreadsheetNotFound:
    print(f"エラー: スプレッドシートが見つかりません。ID '{spreadsheet_id}' が正しいか、アクセス権があるか確認してください。")
except gspread.exceptions.WorksheetNotFound:
    print(f"エラー: シート '{SHEET_NAME}' が見つかりません。スプレッドシート内にその名前のシートが存在するか確認してください。")
except Exception as e:
    print(f"予期せぬエラーが発生しました: {type(e).__name__} ({e})")
    import traceback
    traceback.print_exc()

# === ▼▼▼ 最終的なトークン計算結果を表示 ▼▼▼ ===
print("\n--- トークン使用量の合計 ---")
print(f"実際に消費した合計入力トークン: {total_input_tokens}")
print(f"実際に消費した合計出力トークン: {total_output_tokens}")
print(f"消費した総合計トークン: {total_input_tokens + total_output_tokens}")

if is_cache_enabled and successful_api_calls > 0:
    total_saved_tokens = cached_content_token_count * successful_api_calls
    print("\n--- コンテキストキャッシュによる節約効果 ---")
    print(f"節約できた合計入力トークン (推定): {total_saved_tokens}")
else:
    print("\nコンテキストキャッシュは使用されませんでした。")
# === ▲▲▲ 最終的なトークン計算結果を表示 ▲▲▲ ===

print("\n--- 全ての処理が終了しました ---")
```
